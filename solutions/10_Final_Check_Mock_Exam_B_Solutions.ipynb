### Question 1

Consider `p = Pauli('XYZ').dot(Pauli('YYX'))`. Which statements are true? (Select two.)

A. `p.to_label()` returns `'ZIY'`

B. `p.phase` is `2`

C. `p[1]` is `Pauli('Y')`

D. `p == Pauli('-iZIY')` is `True`

E. `p.equiv(Pauli('ZIY'))` is `True`

Answer: B, E

`dot` is the matrix product, taken qubit by qubit: qubit 2 gives X·Y = iZ, qubit 1 gives Y·Y = I and qubit 0 gives Z·X = iY. So p = i·i·(Z ⊗ I ⊗ Y) = −ZIY. Qiskit stores the coefficient as (−i)^phase, and −1 = (−i)^2, so `phase` is 2. `to_label()` keeps the sign and returns '-ZIY' (A), `p[1]` is the identity (C), and the coefficient is −1, not −i (D). `equiv` ignores the phase, so E is true.

### Question 6

With gate-folding zero-noise extrapolation, ⟨O⟩ is measured as 0.60 at noise factor 1 and as 0.40 at noise factor 3. What zero-noise estimate does a linear extrapolator return?

A. 0.80

B. 0.50

C. 0.70

D. 0.60

Answer: C

The line through (1, 0.60) and (3, 0.40) has a slope of −0.10 per unit of noise factor. Extrapolating back one unit to noise factor 0 gives 0.60 + 0.10 = 0.70. 0.80 wrongly extrapolates by the full 1→3 difference, and 0.50 is just the mean of the two points. 0.60 is the unmitigated value at factor 1, which the `"fallback"` extrapolator would return.

### Question 9

```python
estimator = EstimatorV2(mode=backend, options={"default_precision": 0.01})
job = estimator.run([(isa, iobs, None, 0.005)], precision=0.02)
print(job.result()[0].metadata["target_precision"])
```

`isa` has no parameters and `iobs` is a layout-matched observable. What is printed?

A. `0.01`

B. `0.02`

C. `0.005`

D. `0.015625`

Answer: C

The sampling budget is resolved in this order:
1. A precision inside the PUB.
2. The `precision` argument of `run()`, for PUBs without their own value. PUB and `run()` values are part of the primitive interface and are always obeyed.
3. On Runtime, an explicit twirling product `num_randomizations × shots_per_randomization`.
4. `options.default_shots`.
5. `options.default_precision`.
6. The service default of 0.015625.

Here the PUB's 0.005 wins. Because shots ≈ 1/precision², that target costs about 40,000 shots, four times the 10,000 needed for 0.01. The `None` in the PUB means there are no parameter values.

```python
from qiskit import QuantumCircuit, qasm3
from qiskit.circuit.classical import expr, types

qc = QuantumCircuit(2, 2)
n = qc.add_var("n", expr.lift(5, types.Uint(8)))
qc.store(n, expr.bit_and(n, 3))
with qc.if_test(expr.equal(n, 1)):
    qc.x(1)
qc.measure([0, 1], [0, 1])
print(qasm3.dumps(qc))
```

Which statement about the exported program is correct?

A. `n` is exported as `input uint[8] n;`, so its value must be supplied when the program runs.

B. The exporter raises an error, because OpenQASM 3 has no unsigned integer type.

C. The program declares `uint[8] n;` and assigns `n = 5;`. After `n = n & 3;` the value is 1, so the `if (n == 1)` block flips `q[1]`.

D. The program declares `const uint[8] n = 5;`, so the later assignment `n = n & 3;` is invalid.

Answer: C

`add_var` creates a local classical variable. A Sampler run with 900 shots returns the outcome `11` in 90 shots, so the estimated probability is p̂ = 0.10. What is the approximate standard error of p̂?

A. 0.033

B. 0.010

C. 0.0001

D. 0.017

Answer: B

An outcome frequency is a binomial proportion, with standard error sqrt(p(1-p)/N) = sqrt(0.1 × 0.9 / 900) = sqrt(0.0001) = 0.010. 0.033 is 1/sqrt(900), which ignores the p(1-p) factor. 0.017 is the value you would get at p = 0.5. 0.0001 is the variance, not the standard error.
